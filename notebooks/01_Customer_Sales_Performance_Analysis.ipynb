# Segment 1 - Customer & Sales Performance Analysis

## Analysis 01 — Top Customers by Revenue
### Business question: 
### Which 10 customers generated the highest total completed-order revenue, and how many completed orders did each place?

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    COUNT(o.order_id) AS total_orders,
    ROUND(SUM(o.order_amount), 2) AS total_revenue
FROM retailops360.source.customers AS c
INNER JOIN retailops360.source.orders AS o
    ON c.customer_id = o.customer_id
WHERE o.order_status = 'Completed'
GROUP BY
    c.customer_id,
    c.customer_name
ORDER BY total_revenue DESC
LIMIT 10;
```

## Analysis 02 — Customer Activation Gap
### Business question: 
### Which registered customers have never placed an order, and how long have they been registered?

```python
%sql
SELECT
    c.customer_id,
    c.customer_name,
    c.signup_date,
    DATEDIFF(CURRENT_DATE(), c.signup_date)
        AS days_since_registration
FROM retailops360.source.customers AS c
LEFT ANTI JOIN retailops360.source.orders AS o
    ON c.customer_id = o.customer_id
ORDER BY days_since_registration DESC;
```

## Analysis 03 — Customer Retention & Purchase Inactivity
### Business question: 
### Which customers made completed purchases in 2025 but have not made any completed purchases in 2026?

```python
%sql

WITH customers_2025 AS (
    SELECT
        c.customer_id,
        c.customer_name
    FROM retailops360.source.customers AS c
    LEFT SEMI JOIN retailops360.source.orders AS o
        ON c.customer_id = o.customer_id
       AND o.order_date >= DATE '2025-01-01'
       AND o.order_date < DATE '2026-01-01'
       AND o.order_status = 'Completed'
)
SELECT
    c.customer_id,
    c.customer_name
FROM customers_2025 AS c
LEFT ANTI JOIN retailops360.source.orders AS o
    ON c.customer_id = o.customer_id
   AND o.order_date >= DATE '2026-01-01'
   AND o.order_date < DATE '2027-01-01'
   AND o.order_status = 'Completed'
ORDER BY c.customer_id;
```
